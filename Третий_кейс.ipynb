### **Посчитайте Retention 15 дня (в процентах) для пользователей, зарегистрированных в январе**

Cохраните результат в переменную `retention_15_day`

**Примечание:** результат округлите до 5 знаков после запятой

```python
# Ваше решение
from datetime import datetime

users = {}
with open('registrations.csv', 'r') as reg:
  title = reg.readline().strip().split(';')
  for i in reg:
    line = i.strip().split(';')
    if (datetime.strptime(line[1], '%Y-%m-%d')).month == 1:
      users[line[0]] = [line[1]]
with open('entries.csv', 'r') as entr:
  title2 = entr.readline().strip().split(';')
  for i in entr:
    line = i.strip().split(';')
    if line[0] in users:
      users[line[0]].append(line[1])

all_users = len(users)
count = 0

for keys, values in users.items():
  for j in values:
    if (datetime.strptime(j, '%Y-%m-%d') - datetime.strptime(values[0], '%Y-%m-%d')).days  == 15:
      count += 1
      break

retention_15_day = round((count/all_users)*100, 5)

print(all_users, count, retention_15_day)
```

### **Посчитайте Lifetime по всем пользователям, посчитанный как интеграл от n-day retention**

Сохраните результат в переменную `lifetime`

**Примечание:** результат округлите до 5 знаков после запятой

```python
# Ваше решение
from datetime import datetime

users = {}
with open('registrations.csv', 'r') as reg:
  title = reg.readline().strip().split(';')
  for i in reg:
    line = i.strip().split(';')
    users[line[0]] = [datetime.strptime(line[1], '%Y-%m-%d').date()]
with open('entries.csv', 'r') as entr:
  title2 = entr.readline().strip().split(';')
  for i in entr:
    line = i.strip().split(';')
    if line[0] in users:
      entry_date = datetime.strptime(line[1], '%Y-%m-%d').date()
      reg_date = users[line[0]][0]
      n_day = (entry_date - reg_date).days
      if n_day not in users[line[0]]:
        users[line[0]].append(n_day)

all_users = len(users)
res = [0] * 31
lifetime = 0

for keys, values in users.items():
  for j in values[1:]:
    if j <= 30:
      res[j] += 1

for i in res:
  retention = i/all_users
  lifetime += retention


lifetime = round(lifetime, 5)

print(lifetime)
```

```python
#@title ✏️ Проверка: чтобы проверить свое решение запустите код в этой ячейке
correct_answer = float(answers[2])

try:
    assert lifetime == correct_answer
except AssertionError:
    print('Ответы не совпадают')
else:
    print('Поздравляем, Вы справились!')
```
